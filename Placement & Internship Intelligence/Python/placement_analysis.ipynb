# Campus Placement Data Analysis
### Beginner Exploratory Data Analysis (EDA) using Python & Pandas

This notebook demonstrates simple data exploration and validation on the cleaned campus placement dataset.

## 1. Load the Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load clean placement data
df = pd.read_csv("../Data/placement_data.csv")
df.head()
```

## 2. Dataset Shape & Summary

```python
# Check number of rows and columns
print("Dataset Shape (Rows, Columns):", df.shape)
df.info()
```

## 3. Data Quality Checks (Missing Values & Duplicates)

```python
# Check for duplicate student records
print("Total Duplicates:", df.duplicated(subset=["Student_ID"]).sum())

# Check for missing values
# Note: Unplaced students naturally have missing values for Company, Job_Role, Salary_LPA, and Location
df.isnull().sum()
```

## 4. Descriptive Statistics

```python
# Numerical summary for CGPA and Salary
df[["CGPA", "Salary_LPA"]].describe()
```

## 5. Category Value Counts

```python
print("--- Department Distribution ---")
print(df["Department"].value_counts())

print("
--- Placement Status ---")
print(df["Placement_Status"].value_counts())

print("
--- Internship Experience ---")
print(df["Internship"].value_counts())
```

## 6. Groupby Analysis

```python
# Department-wise Placement Rate & Average Salary
dept_summary = df.groupby("Department").agg(
    total_students=("Student_ID", "count"),
    placed_students=("Placement_Status", lambda x: (x == "Placed").sum()),
    avg_salary=("Salary_LPA", "mean")
).reset_index()

dept_summary["placement_rate_pct"] = (dept_summary["placed_students"] / dept_summary["total_students"]) * 100
dept_summary.sort_values(by="placement_rate_pct", ascending=False)
```

```python
# Internship vs Placement Status
intern_summary = df.groupby("Internship").agg(
    total=("Student_ID", "count"),
    placed=("Placement_Status", lambda x: (x == "Placed").sum()),
    avg_salary=("Salary_LPA", "mean")
).reset_index()

intern_summary["placement_rate_pct"] = (intern_summary["placed"] / intern_summary["total"]) * 100
intern_summary
```

```python
# Top 5 Hiring Companies
df[df["Placement_Status"] == "Placed"]["Company"].value_counts().head(5)
```

## 7. Simple Visualization

```python
# Simple bar chart of Department Placement Rates
plt.figure(figsize=(8, 4))
plt.bar(dept_summary["Department"], dept_summary["placement_rate_pct"], color="#3b82f6")
plt.title("Placement Rate by Department (%)")
plt.ylabel("Placement Rate (%)")
plt.xticks(rotation=20, ha="right")
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```
